## 2. Tensor Pertama dengan Contoh Data Cuaca Mini

Semua data yang masuk ke jaringan saraf direpresentasikan sebagai **tensor** — generalisasi matriks ke dimensi berapa pun.

Di bawah: contoh suhu minimum & maksimum harian.

```python
import tensorflow as tf

# Tensor 1D: suhu harian, 3 pengamatan
suhu_hari = tf.constant([26.5, 26.8, 27.2])

print("Data:", suhu_hari)
print("Bentuk (shape):", suhu_hari.shape)
print("Dimensi (rank):", tf.rank(suhu_hari).numpy())
```

```python
# Tensor 2D: suhu minimum & maksimum selama 3 hari
suhu_3_hari = tf.constant([
    [26.5, 31.0],
    [26.8, 30.5],
    [27.2, 32.1],
])

print("Bentuk (shape):", suhu_3_hari.shape)
print("Dimensi (rank):", tf.rank(suhu_3_hari).numpy())
```

```python
# Tensor 3D: suhu minimum & maksimum di 3 stasiun selama 3 hari
# Bentuk: [hari, stasiun, variabel]
suhu_3_stasiun = tf.constant([
    [[26.5, 31.0], [26.1, 30.7], [27.0, 32.4]],
    [[26.8, 30.5], [26.4, 30.2], [27.3, 31.9]],
    [[27.2, 32.1], [26.9, 31.8], [27.8, 32.8]],
])

print("Bentuk (shape):", suhu_3_stasiun.shape)
print("Dimensi (rank):", tf.rank(suhu_3_stasiun).numpy())
```

## 3. Operasi Sederhana pada Tensor (Teaser)

Tensor dapat dioperasikan: penjumlahan, perkalian, dan sebagainya. Bab 4 akan membahas bagaimana tensor dipakai untuk melatih model neural network.

```python
import tensorflow as tf

# Contoh operasi sederhana: perkalian elementwise
a = tf.constant([1.0, 2.0, 3.0])
b = tf.constant([2.0, 2.0, 2.0])

print("a * b =", a * b)
```
